```python
%matplotlib inline

import matplotlib.pyplot as plt
from copy import deepcopy
import numpy as np
import pystan 
import pandas as pd
import pickle
from tqdm.notebook import tqdm as tqdm
```

```python
datum = pd.read_csv('../../X data/Chicken2Hand.csv')
pd.set_option('display.max_rows', None)
print(datum)
```

```python
datum = datum.drop(2)
```

```python
model = '''
data {
    int N;  // number of measurement cell
    real LogNb[N];  // before
    real LogNa[N];  // after
}

transformed data{
    real <lower = 0, upper =1> p_cross[N];
    
    for (i in 1:N){ 
        p_cross[i]  = 10^(LogNa[i] - LogNb[i]);
    }
}

parameters {
    real <lower = 0> alpha;
    real <lower = 0> beta;
}

model{
    for (i in 1:N){ 
        p_cross[i] ~ beta(alpha, beta);
    }
}

'''

sm = pystan.StanModel(model_code=model)
```

```python
fit_NUTS = sm.sampling(
    data = {'N' : len(datum["Chicken"]), 'LogNb' : datum["Chicken"], 'LogNa' : datum["Hand"]},
    iter = 2500+5000,
    chains = 4,
    warmup = 5000,
    thin = 10,
    seed = 1234
)
```

```python
print(fit_NUTS)
```
